```python
# envelope函数
import numpy as np
import matplotlib.pyplot as plt

# Define the polynomial coefficients
a = 1
b = -21
c = 35
d = -7.5
exponent = 7

# Generate x values
x = np.linspace(0, 1, 1000)

# Calculate the polynomial values
# y = a + b * x**5 + c * x**6 + d * x**7

y = np.exp(-(x**2)/(1-x**2))

# Plot the polynomial
plt.figure(figsize=(8, 6), facecolor='white')
plt.plot(x, y, label='1 - 21x^5 + 35x^6 - 7.5x^7', color='blue')
plt.title('Polynomial Function')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```
